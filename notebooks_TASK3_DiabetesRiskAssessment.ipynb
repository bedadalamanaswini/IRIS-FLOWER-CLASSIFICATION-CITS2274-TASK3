```python
import pandas as pd
from sklearn.linear_model import LogisticRegression

# Dataset
data = {
    "Age": [25,45,35,50,30,60,40,55],
    "Glucose": [90,180,120,200,100,220,140,190],
    "BMI": [22,35,28,40,24,45,30,38],
    "Diabetes": [0,1,0,1,0,1,1,1]
}

df = pd.DataFrame(data)

# Input and output
X = df[["Age","Glucose","BMI"]]
y = df["Diabetes"]

# Train model
model = LogisticRegression()
model.fit(X, y)

print("Model Training Completed")
```

```python
# Predict new person

person = pd.DataFrame(
    [[50,170,32]],
    columns=["Age","Glucose","BMI"]
)

prediction = model.predict(person)

print("Prediction Result:", prediction)

if prediction[0] == 1:
    print("High Diabetes Risk")
else:
    print("Low Diabetes Risk")
```
